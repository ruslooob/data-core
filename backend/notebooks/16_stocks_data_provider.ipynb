# Практическое использование StocksDataCore

```python
from core.stock_data_provider import StockDataProvider
_stocks = StockDataProvider()
from core.plot_utils import plot_2d

_stocks.list_tickers()[:10]
```

```python
_stocks.get_ticker_info('VTBR')
```

```python
_stocks.get_ticker_info('VTBR', start_date='2020-01-01', end_date='2022-12-31')
```

```python
ticker = 'VTBR'
stock_data = _stocks.get_candles(ticker)

plot_2d(
    x=stock_data["DATE"],
    y=stock_data["CLOSE"],
    xlabel="Дата",
    ylabel="Цена закрытия (₽)",
    title=f"Динамика акций {ticker} (нормализованная)"
)
```

```python
stock_data_raw = _stocks.get_candles(ticker, normalized=False)

plot_2d(
    x=stock_data_raw["DATE"],
    y=stock_data_raw["CLOSE"],
    xlabel="Дата",
    ylabel="Цена закрытия (₽)",
    title=f"Динамика акций {ticker} (без учёта сплитов)"
)
```

```python
stock_data_filtered = _stocks.get_candles(ticker, start_date='2020-01-01', end_date='2022-12-31')

plot_2d(
    x=stock_data_filtered["DATE"],
    y=stock_data_filtered["CLOSE"],
    xlabel="Дата",
    ylabel="Цена закрытия (₽)",
    title=f"Динамика акций {ticker} (2020–2022)"
)
```
