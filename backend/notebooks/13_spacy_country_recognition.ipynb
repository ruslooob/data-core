```python
import pandas as pd
import time
from core.spacy_country_recognition import CountryRecognizer
```

```python
df = pd.read_csv('../../data/events/3_countries/2000-2025.csv')

df.head()
```

```python
recognizer = CountryRecognizer()

start = time.time()
df["country"] = recognizer.match_locations_series(df["event"])
elapsed = time.time() - start

recognized = df["country"].notna().sum()
print(f"Время:     {elapsed:.2f} с")
print(f"Опознано:  {recognized} / {len(df)} строк ({recognized / len(df) * 100:.1f}%)")
df.head(10)
```
