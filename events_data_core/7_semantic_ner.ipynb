```python
import pandas as pd
from gensim.models import KeyedVectors
```

```python
df = pd.read_csv('../events/2_struct/2000-2025.csv')

df.head()
```

```python
path = "../models/cc.ru.300.vec.gz"

model = KeyedVectors.load_word2vec_format(path, binary=False)

print(model.similarity("россия", "москва"))     # должно быть высоко (около 0.6–0.8)
print(model.similarity("россия", "америка"))    # ниже, чем с "москва"
print(model.most_similar("россия", topn=20))     # ближайшие слова к "россия"
```
